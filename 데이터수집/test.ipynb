# 외감기업 (코스닥 + 코스피 + 비상장 외감기업)

```python
import pandas as pd

cf1 = pd.read_excel('TS2000외감_2024(1).xlsx')
cf2 = pd.read_excel('TS2000외감_2024(2).xlsx')
cf3 = pd.read_excel('TS2000외감_2024(3).xlsx')

df = pd.concat([cf1,cf2,cf3], ignore_index=True)
```

```python
# 대분류 안에 각 고윳값(=대분류 섹터)의 비중 확인

( df['[공통]통계청 한국표준산업분류 11차(대분류)']
    .value_counts(normalize=True)
    * 100
).round(2)
```

```python
( df['[공통]통계청 한국표준산업분류 11차(중분류)']
    .value_counts(normalize=True)
    * 100
).round(2)
```

```python
result = (
    df.groupby(
        ['[공통]통계청 한국표준산업분류 11차(대분류)',
         '[공통]통계청 한국표준산업분류 11차(중분류)']
    )
    .size()
    .reset_index(name='중분류 기업수')
)


total_n = len(df)

# 중분류가 전체 데이터에서 차지하는 비중 (%)
result['중분류_전체비중(%)'] = (
    result['중분류 기업수'] / total_n * 100
).round(2)

# 대분류별 중분류 개수
result['대분류 안에 중분류개수'] = (
    result.groupby('[공통]통계청 한국표준산업분류 11차(대분류)')
    ['[공통]통계청 한국표준산업분류 11차(중분류)']
    .transform('nunique')
)


# 대분류 전체 기업수
result['대분류_전체기업수'] = (
    result.groupby('[공통]통계청 한국표준산업분류 11차(대분류)')['중분류 기업수']
    .transform('sum')
)

# 대분류 비중 (%)
result['대분류비중(%)'] = (
    result['대분류_전체기업수'] / total_n * 100
).round(2)


# 대분류 비중 큰 순으로 정렬
result = result.sort_values(
    by=['중분류 기업수'],
    ascending=False
)

result
```

# 전체비중이 2% 넘는 중분류에 대해서 분석시 (17개 분류로 나뉘어진다)

```python
result.loc[
    result['중분류_전체비중(%)'] >= 2,
    [
        '[공통]통계청 한국표준산업분류 11차(대분류)',
        '[공통]통계청 한국표준산업분류 11차(중분류)',
        '중분류_전체비중(%)'
    ]
]
```

```python
# 이경우 비상장외감기업 + 코스닥 상장기업의 72%를 분석할 수 있다
result.loc[
    result['중분류_전체비중(%)'] >= 2,
    '중분류_전체비중(%)'
].sum()
```
