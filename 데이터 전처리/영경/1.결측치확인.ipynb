```python
import pandas as pd
import numpy as np
from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill, Alignment

df = pd.read_csv(r'C:\유비온프로젝트2\corporate-bankruptcy\데이터 전처리\영경\M23_부동산_임대_사업서비스업_정제.csv', encoding='utf-8-sig')

print("=== 데이터 기본 정보 ===")
print(f"행 수: {df.shape[0]}, 열 수: {df.shape[1]}")
print()

# 결측치 분석
missing_count = df.isnull().sum()
missing_ratio = (df.isnull().sum() / len(df) * 100).round(2)

# 0값 분석 (숫자형 컬럼만)
numeric_cols = df.select_dtypes(include=[np.number]).columns
zero_count = (df[numeric_cols] == 0).sum()
zero_ratio = (zero_count / len(df) * 100).round(2)

# 결측치 요약 테이블
missing_df = pd.DataFrame({
    '결측치 수': missing_count,
    '결측치 비율(%)': missing_ratio
})
missing_df = missing_df[missing_df['결측치 수'] > 0].sort_values('결측치 비율(%)', ascending=False)

print("=== 결측치 현황 (결측치가 있는 컬럼만) ===")
if missing_df.empty:
    print("결측치 없음")
else:
    print(missing_df.to_string())
print()

# 0값 요약 테이블
zero_df = pd.DataFrame({
    '0값 수': zero_count,
    '0값 비율(%)': zero_ratio
})
zero_df = zero_df[zero_df['0값 수'] > 0].sort_values('0값 비율(%)', ascending=False)

print("=== 0값 현황 (숫자형 컬럼, 0값이 있는 컬럼만) ===")
if zero_df.empty:
    print("0값 없음")
else:
    print(zero_df.to_string())
print()

# 결측치 + 0값 통합 요약 (숫자형 컬럼 기준)
print("=== 결측치 + 0값 통합 현황 (숫자형 컬럼) ===")
combined_df = pd.DataFrame({
    '결측치 수': missing_count[numeric_cols],
    '결측치 비율(%)': missing_ratio[numeric_cols],
    '0값 수': zero_count,
    '0값 비율(%)': zero_ratio,
})
combined_df['결측+0값 수'] = combined_df['결측치 수'] + combined_df['0값 수']
combined_df['결측+0값 비율(%)'] = (combined_df['결측+0값 수'] / len(df) * 100).round(2)
combined_df = combined_df.sort_values('결측+0값 비율(%)', ascending=False)
print(combined_df.to_string())

# ────────────────────────────────────────────
# 엑셀 저장
# ────────────────────────────────────────────
output_path = r'C:\유비온프로젝트2\corporate-bankruptcy\데이터 전처리\영경\결측치_0값_분석.xlsx'

def write_sheet(ws, df_data, title):
    """데이터프레임을 시트에 작성하고 헤더 서식 적용"""
    header_font = Font(name='Arial', bold=True, color='FFFFFF')
    header_fill = PatternFill('solid', start_color='2F5496')
    center = Alignment(horizontal='center')

    # 타이틀 행
    ws.append([title])
    ws['A1'].font = Font(name='Arial', bold=True, size=12)
    ws.append([])  # 빈 줄

    # 헤더
    headers = ['컬럼명'] + list(df_data.columns)
    ws.append(headers)
    for col_idx, _ in enumerate(headers, start=1):
        cell = ws.cell(row=3, column=col_idx)
        cell.font = header_font
        cell.fill = header_fill
        cell.alignment = center

    # 데이터
    for row in df_data.reset_index().itertuples(index=False):
        ws.append(list(row))

    # 열 너비 자동 조정
    for col in ws.columns:
        max_len = max((len(str(c.value)) if c.value else 0) for c in col)
        ws.column_dimensions[col[0].column_letter].width = min(max_len + 4, 40)


wb = Workbook()

# 시트 1: 결측치 현황
ws1 = wb.active
ws1.title = '결측치 현황'
if missing_df.empty:
    ws1.append(['결측치 없음'])
else:
    write_sheet(ws1, missing_df, '결측치 현황 (결측치가 있는 컬럼만)')

# 시트 2: 0값 현황
ws2 = wb.create_sheet('0값 현황')
if zero_df.empty:
    ws2.append(['0값 없음'])
else:
    write_sheet(ws2, zero_df, '0값 현황 (0값이 있는 숫자형 컬럼만)')

# 시트 3: 통합 현황
ws3 = wb.create_sheet('결측+0값 통합')
write_sheet(ws3, combined_df, '결측치 + 0값 통합 현황 (숫자형 컬럼)')

wb.save(output_path)
print(f"\n엑셀 저장 완료: {output_path}")
```

## 결측치 행 확인

```python
# 결측치가 있는 행 확인
missing_rows = df[df.isnull().any(axis=1)]
print(f"결측치 있는 행 수: {len(missing_rows)}")
print(f"전체 행 대비 비율: {len(missing_rows)/len(df)*100:.2f}%")

# 해당 행들이 어떤 컬럼에서 결측인지 확인
print(missing_rows.isnull().sum(axis=1).value_counts())  # 행별 결측 컬럼 수 분포

# 결측 행의 레이블(부도여부) 분포 확인 — 컬럼명은 실제에 맞게 수정
if '부도여부' in df.columns:
    print(missing_rows['부도여부'].value_counts())
```
