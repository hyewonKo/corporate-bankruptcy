## 연도별로 자산총계 top10 기업 확인

```python
import pandas as pd

df = pd.read_parquet(r'C:\유비온프로젝트2\corporate-bankruptcy\데이터 전처리\데이터\7번\M19_도매_소매업.parquet')

print(df.shape)

# 연도별 자산총계 상위 10개 기업 추출
result = (
    df.groupby('회계년도')
    .apply(lambda x: x.nlargest(10, '자산총계(요약)(백만원)')[['회사명', '자산총계(요약)(백만원)']])
    .reset_index(level=1, drop=True)
    .reset_index()
)

# 순위 컬럼 추가
result['순위'] = result.groupby('회계년도').cumcount() + 1
result = result[['회계년도', '순위', '회사명', '자산총계(요약)(백만원)']]

# CSV 저장
result.to_parquet(r'C:\유비온프로젝트2\corporate-bankruptcy\데이터 전처리\데이터\8,9번.자산총계top10\M19_도매_소매업_top10.parquet', index=False)
print("저장 완료: top10_자산총계_연도별.csv")

# 연도별로 출력
for year, group in result.groupby('회계년도'):
    print(f"\n── {year}년 ──")
    print(group.to_string(index=False))
```

## 연도별 top10 기업 데이터 평균

```python
import pandas as pd

# 데이터 로드
df = pd.read_parquet(r'C:\유비온프로젝트2\corporate-bankruptcy\데이터 전처리\데이터\7번\M19_도매_소매업.parquet')
top10 = pd.read_parquet(r'C:\유비온프로젝트2\corporate-bankruptcy\데이터 전처리\데이터\8,9번.자산총계top10\M19_도매_소매업_top10.parquet')

# 전체 데이터에서 연도별 top10 기업 행만 필터링
filtered = df.merge(top10[['회계년도', '회사명']], on=['회계년도', '회사명'], how='inner')

# 숫자형 컬럼만 평균 계산 (회계년도는 groupby 키이므로 제외)
exclude = ['회계년도', '사업자등록번호', '회사명', '통계청 한국표준산업분류 코드 11차(대분류)', '통계청 한국표준산업분류 11차(중분류)', 'M코드']
num_cols = [c for c in filtered.select_dtypes(include='number').columns if c not in exclude]

# 연도별 평균 (13행 × 69컬럼)
result = filtered.groupby('회계년도')[num_cols].mean().reset_index()

# CSV 저장
result.to_parquet(r'C:\유비온프로젝트2\corporate-bankruptcy\데이터 전처리\데이터\8,9번.자산총계top10평균\M19_도매_소매업_top10평균.parquet', index=False)
print(f"저장 완료: top10_연도별_재무지표_평균.csv")
print(f"shape: {result.shape}  ({result.shape[0]}개 연도 × {result.shape[1]}개 컬럼)")
print(result[['회계년도', 'ROA', '영업이익률', '부채비율']].to_string(index=False))
```

## 본래 데이터에 산업top10 칼럼 추가하기

```python
import pandas as pd

df = pd.read_parquet(r'C:\유비온프로젝트2\corporate-bankruptcy\데이터 전처리\데이터\7번\M19_도매_소매업.parquet')
top10 = pd.read_parquet(r'C:\유비온프로젝트2\corporate-bankruptcy\데이터 전처리\데이터\8,9번.자산총계top10평균\M19_도매_소매업_top10평균.parquet')

# top10이 연도별 1행인지 먼저 검증
assert top10['회계년도'].nunique() == len(top10), "top10 파일에 연도 중복 있음 → merge 전 중단"

# 식별 컬럼 (연산 제외)
id_cols = ['사업자등록번호', '회계년도', '회사명',
           '통계청 한국표준산업분류 코드 11차(대분류)',
           '통계청 한국표준산업분류 11차(중분류)', 'M코드']

# 연산 대상: top10 평균 파일과 공통인 숫자형 컬럼
target_cols = [c for c in top10.columns if c != '회계년도' and c in df.columns]

# 연도별 top10 평균을 원본 df에 left join
avg = top10[['회계년도'] + target_cols].rename(columns={c: c + '_avg' for c in target_cols})
df_merged = df.merge(avg, on='회계년도', how='left').reset_index(drop=True)  # ← 인덱스 리셋

# merge 후 행 수 검증
assert len(df_merged) == len(df), f"행 폭발 발생! 원본: {len(df)}행 → merge 후: {len(df_merged)}행"

# diff / ratio 컬럼 생성 (.values로 인덱스 무관하게 연산)
extra = pd.DataFrame(
    {col + '_diff' : df_merged[col].values - df_merged[col + '_avg'].values for col in target_cols}
    | {col + '_ratio': df_merged[col].values / df_merged[col + '_avg'].replace(0, pd.NA).values for col in target_cols}
)

ordered_extra = [suffix for col in target_cols for suffix in (col + '_diff', col + '_ratio')]
extra = extra[ordered_extra]

# 원본 + 추가 컬럼 합치기 (인덱스 맞춰서 concat)
result = pd.concat([df.reset_index(drop=True), extra], axis=1)

# 최종 행 수 검증
assert len(result) == len(df), f"최종 행 폭발! 원본: {len(df)}행 → 결과: {len(result)}행"

print(f'shape: {result.shape}  ({result.shape[0]}행 × {result.shape[1]}컬럼)')
print(f'  - 원본 컬럼: {len(df.columns)}개')
print(f'  - 추가 컬럼: {len(extra.columns)}개 (target {len(target_cols)}개 × diff/ratio 2종)')

result.to_parquet(
    r'C:\유비온프로젝트2\corporate-bankruptcy\데이터 전처리\데이터\8,9번.withTOP10\M19_도매_소매업_withTOP10.parquet',
    index=False
)
print('저장 완료')
```

```python
import pandas as pd
import numpy as np

# ── 1. M19 기업 데이터 로드
df = pd.read_parquet(r'C:\유비온프로젝트2\corporate-bankruptcy\데이터 전처리\데이터\7번\M19_도매_소매업.parquet')

# ── 2. 산업평균(코드 G) 데이터 로드
industry_files = [
    r'C:\유비온프로젝트2\corporate-bankruptcy\데이터수집\산업평균(에코스)\손익_지표_처리결과.xlsx',
    r'C:\유비온프로젝트2\corporate-bankruptcy\데이터수집\산업평균(에코스)\자산_자본_지표_처리결과.xlsx',
    r'C:\유비온프로젝트2\corporate-bankruptcy\데이터수집\산업평균(에코스)\자산_자본_회전율_지표_처리결과.xlsx',
    r'C:\유비온프로젝트2\corporate-bankruptcy\데이터수집\산업평균(에코스)\성장성_지표_처리결과.xlsx',
]

NEED_ITEMS = [
    'EBITDA대매출액', 'EBIT대매출액', '금융비용대매출액', '매출액순이익률',
    '매출액영업이익률', '매출원가대매출액', '사내유보율',
    '당좌비율', '부채비율', '비유동비율', '비유동장기적합률',
    '순운전자본대총자본', '유동비율', '자기자본비율', '차입금의존도', '현금비율',
    '매입채무회전율', '매출채권회전율', '비유동자산회전율', '유형자산회전율',
    '자기자본회전율', '재고자산회전율', '총자산회전율',
    '매출액증가율', '유형자산증가율', '자기자본증가율', '총자산증가율',
]

raw_dfs = []
for path in industry_files:
    xdf = pd.read_excel(path)
    g = xdf[xdf['코드(업종코드)'].astype(str).str.strip() == 'G']
    g = g[g['계정항목'].isin(NEED_ITEMS)].copy()
    raw_dfs.append(g)

industry_raw = pd.concat(raw_dfs, ignore_index=True)

# 중복 항목 확인
dupes = industry_raw.groupby('계정항목').size()
if (dupes > len([str(y) for y in range(2012, 2025)])).any():
    print("⚠️  중복 계정항목 존재:", dupes[dupes > 13].index.tolist())

# wide → long → pivot
year_cols = [str(y) for y in range(2012, 2025) if str(y) in industry_raw.columns]
industry_long = industry_raw.melt(
    id_vars=['계정항목'], value_vars=year_cols,
    var_name='회계년도', value_name='산업평균'
)
industry_long['회계년도'] = industry_long['회계년도'].astype(int)
industry_pivot = industry_long.pivot_table(
    index='회계년도', columns='계정항목', values='산업평균',
    aggfunc='first'   # ← 중복 시 mean 대신 first로 명시
).reset_index()
industry_pivot.columns.name = None

rename_map = {c: c + '_avg' for c in industry_pivot.columns if c != '회계년도'}
industry_pivot = industry_pivot.rename(columns=rename_map)

# ── 3. join 후 인덱스 리셋 ← 핵심 수정
df_merged = df.merge(industry_pivot, on='회계년도', how='left').reset_index(drop=True)

assert len(df_merged) == len(df), f"행 폭발! 원본: {len(df)}행 → merge 후: {len(df_merged)}행"

# ── 4. 매핑 정의
MAPPING = [
    ('영업이익률',       '매출액영업이익률_avg', '매출액영업이익률'),
    ('영업이익률',       'EBIT대매출액_avg',     'EBIT대매출액'),
    ('EBITDA마진',      'EBITDA대매출액_avg',   'EBITDA대매출액'),
    ('매출원가율',       '매출원가대매출액_avg', '매출원가대매출액'),
    ('순이익률',         '매출액순이익률_avg',   '매출액순이익률'),
    ('유보율',           '사내유보율_avg',       '사내유보율'),
    ('금융비용부담률',   '금융비용대매출액_avg', '금융비용대매출액'),
    ('자기자본비율',     '자기자본비율_avg',      '자기자본비율'),
    ('부채비율',         '부채비율_avg',          '부채비율'),
    ('유동비율',         '유동비율_avg',          '유동비율'),
    ('당좌비율_추정',    '당좌비율_avg',          '당좌비율'),
    ('현금비율',         '현금비율_avg',          '현금비율'),
    ('순운전자본비율',   '순운전자본대총자본_avg','순운전자본대총자본'),
    ('비유동비율',       '비유동비율_avg',        '비유동비율'),
    ('비유동장기적합률', '비유동장기적합률_avg',  '비유동장기적합률'),
    ('차입금의존도',     '차입금의존도_avg',      '차입금의존도'),
    ('총자산회전율',     '총자산회전율_avg',      '총자산회전율'),
    ('자기자본회전율',   '자기자본회전율_avg',    '자기자본회전율'),
    ('비유동자산회전율', '비유동자산회전율_avg',  '비유동자산회전율'),
    ('유형자산회전율',   '유형자산회전율_avg',    '유형자산회전율'),
    ('재고자산회전율',   '재고자산회전율_avg',    '재고자산회전율'),
    ('매출채권회전율',   '매출채권회전율_avg',    '매출채권회전율'),
    ('매입채무회전율',   '매입채무회전율_avg',    '매입채무회전율'),
    ('총자산증가율',     '총자산증가율_avg',      '총자산증가율'),
    ('유형자산증가율',   '유형자산증가율_avg',    '유형자산증가율'),
    ('자기자본증가율',   '자기자본증가율_avg',    '자기자본증가율'),
    ('매출액증가율',     '매출액증가율_avg',      '매출액증가율'),
]

# ── 5. diff / ratio 컬럼 생성 (.values로 인덱스 무관하게 연산)
id_cols = [
    '사업자등록번호', '회계년도', '회사명', '종업원',
    '통계청 한국표준산업분류 코드 11차(대분류)',
    '통계청 한국표준산업분류 11차(중분류)', 'M코드',
]

extra_cols = {}
for m19_col, avg_col, label in MAPPING:
    extra_cols[f'{label}_diff']  = df_merged[m19_col].values - df_merged[avg_col].values
    extra_cols[f'{label}_ratio'] = df_merged[m19_col].values / df_merged[avg_col].replace(0, pd.NA).values

extra_df = pd.DataFrame(extra_cols)

# concat 시 인덱스 맞춤 ← 핵심 수정
result = pd.concat([df[id_cols].reset_index(drop=True), extra_df], axis=1)

assert len(result) == len(df), f"최종 행 폭발! 원본: {len(df)}행 → 결과: {len(result)}행"
print(f"최종 shape: {result.shape}  ({result.shape[0]}행 × {result.shape[1]}컬럼)")

# ── 6. 저장
result.to_parquet(
    r'C:\유비온프로젝트2\corporate-bankruptcy\데이터 전처리\데이터\8,9번.산업평균비교\M19_도매_소매업_산업평균.parquet',
    index=False
)
print("저장 완료")
```

## 칼럼 합치기

```python
import pandas as pd

left  = pd.read_parquet(r'C:\유비온프로젝트2\corporate-bankruptcy\데이터 전처리\데이터\8,9번.withTOP10\M19_도매_소매업_withTOP10.parquet')
right = pd.read_parquet(r'C:\유비온프로젝트2\corporate-bankruptcy\데이터 전처리\데이터\8,9번.산업평균비교\M19_도매_소매업_산업평균.parquet')

# 행 수 먼저 검증
assert len(left) == len(right), f"행 수 불일치! left: {len(left)}행, right: {len(right)}행"

# (사업자등록번호, 회계년도) 순서 동일한지 검증
key_cols = ['사업자등록번호', '회계년도']
assert (left[key_cols].values == right[key_cols].values).all(), "키 순서 불일치! 정렬 필요"

# diff/ratio 컬럼에 _industry suffix 부여
rename_map = {c: c + '_industry' for c in right.columns if c.endswith('_diff') or c.endswith('_ratio')}
right_renamed = right.rename(columns=rename_map)

# merge 없이 인덱스 기준으로 직접 붙이기
extra_cols = list(rename_map.values())
result = pd.concat([left.reset_index(drop=True), right_renamed[extra_cols].reset_index(drop=True)], axis=1)

assert len(result) == len(left), f"행 폭발! {len(left)} → {len(result)}"
print(f"최종 shape: {result.shape}")

result.to_parquet(
    r'C:\유비온프로젝트2\corporate-bankruptcy\데이터 전처리\데이터\8,9번 파일(최종)\M19_도매_소매업(최종).parquet',
    index=False
)
print("저장 완료")
```
