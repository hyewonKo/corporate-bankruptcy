```python
from pathlib import Path
import shutil

base_dir = Path(r".")
summary_dir = base_dir / "요약"
summary_dir.mkdir(exist_ok=True)

for industry_dir in base_dir.iterdir():
    if not industry_dir.is_dir() or industry_dir.name == "요약":
        continue

    # *_summary.csv 파일 찾기
    summary_files = list(industry_dir.glob("*_summary.csv"))

    if summary_files:
        src_file = summary_files[0]  # 보통 1개라고 가정

        # 산업명으로 파일명 변경하여 저장
        dst_file = summary_dir / f"{industry_dir.name}_summary.csv"

        shutil.copy2(src_file, dst_file)
        print(f"복사 완료: {dst_file.name}")

print("완료")
```
